# FOR LOOP
#### A for loop in Python is used to repeat a block of code by iterating over a sequence (such as a list, tuple, string, dictionary, or set) or any other iterable object
```python
for item in iterable_object:
    # Code block to execute
    # Indentation is mandatory
```
-------------
#### An iterable object is some kind of collection of items: list of numbers, string of char, a range...
#### item is a new variable that can be called whatever you want - basically it represents where we are in the loop 

````python
for char in 'hello':
    print(char)
````

````python
for i in range(0,10):
    print(i*2)
````

````python
for char in 'coffee':
    print(char * 10)
````

### FOR LOOP basic exercise

````python
x = int(input('How many times do I have to tell you? '))
y = ['Clean up your room!']

for i in y:
    print(y * x)
````

````python
for i in range(1,20):

    if i == 4 or i == 13:
        print(f'{i} is unlucky')
    elif i % 2 == 0:
        print(f'{i} is even')
    elif i % 2 != 0:
        print(f'{i} is odd')
````
