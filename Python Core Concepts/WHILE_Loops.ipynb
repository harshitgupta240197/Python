# WHILE LOOPS
#### A while loop is used to repeatedly execute a block of code as long as a specified condition evaluates to True. It is typically used when you do not know the exact number of iterations in advance.
```python
while condition:
    # Code block to execute
    # (Must be indented)

```

````python
for i in range(1,10):
    print('*' * i)
````

````python
counter = 0

while counter <= 10:

    print('*' * counter)
    counter+=1
````

````python
statement = input('Hey how is it going ? ')

while statement != 'stop copying me':
    print(statement)
    statement = input()

print('UGH FINE YOU WIN')
    
````
