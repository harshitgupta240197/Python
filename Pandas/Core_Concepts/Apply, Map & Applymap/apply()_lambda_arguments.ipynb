# apply() with Lambda & Arguments

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
```

```python
titanic['age'] = titanic['age'].replace(['?'], [None]).astype('float')
titanic['fare'] = titanic['fare'].replace(['?'], [None]).astype('float')
```

#### We can make a normal function like this using def

```python
def dolla_rich(money):
    return f'${money*24}'
```

```python
dolla_rich(5)
```

# Or we can instead write the same function as above in one line using LAMDA

```python
titanic['fare'].apply(lambda x: f'${x*24}')
```

# Let us make a currency converter function

```python
def currency_converter(num, multiplier):
    return f'${num * multiplier}'
```

```python
titanic['fare'].apply(currency_converter, args = (24,))
```
