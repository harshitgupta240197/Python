# applymap() method
#### applymap method in pandas applies a function to every single individual element (cell) across an entire DataFrame.

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
```

```python
titanic['age'] = titanic['age'].replace(['?'], [None]).astype('float')
```

```python
def get_age_group(age):
    if age < 2:
        return 'infant'
    elif age < 12:
        return 'child'
    elif age < 18:
        return 'teen'
    elif age < 50:
        return 'adult'
    else:
        return 'senior'
```

```python
titanic['age_group'] = titanic['age'].apply(get_age_group)
```
