# pandas.Series.map()
#### This method is used to substitute each value in a Pandas Series with another value derived from a dictionary, a function, or another Series

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
```

```python
titanic['age'] = titanic['age'].replace(['?'], [None]).astype('float')
```

```python
titanic['pclass']
```

## We will rename 'pclass' data from 1, 2, 3 to 1st, 2nd, 3rd using .map({})

```python
titanic['pclass'].map({1 : '1st', 2 : '2nd', 3 : '3rd'})
```

# Using LAMBDA we will create a function that checks if the age is below 18 or not ?

```python
titanic['age'].map(lambda x: x < 18)
```
