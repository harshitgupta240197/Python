# String vs Object Datatype
### Object is a catch-all data type inherited from NumPy that holds pointers to any Python object, while String is a dedicated data type specifically designed to handle text data.

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
```

```python
titanic.info()
```

```python
titanic['age'] = titanic['age'].replace(['?', None]).astype('float')
titanic['fare'] = titanic['fare'].replace(['?', None]).astype('float')
```

```python
titanic['name'] = titanic['name'].astype('string')
```
