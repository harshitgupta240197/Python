# Stripping Whitespace with strip(), lstrip(), rstrip()
##### Removes any leading or trailing whitespace characters

```python
import pandas as pd
```

```python
s = pd.Series(['1.  Hawk. ', '2.  Avengers  \n', 'watermel  on'])
```

```python
s.str.strip()
```

```python
s.str.strip(to_strip='123.')
```

```python
s.str.strip(to_strip='123. \n \t')
```

```python
s.str.lstrip()
```

```python
s.str.rstrip()
```
