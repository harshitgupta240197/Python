# contains()
### The str.contains() method in pandas tests whether a specific substring or regular expression pattern exists within the text elements of a Series or Index

```python
import pandas as pd
```

```python
ufo = pd.read_csv('../../../data/nuforc_reports.csv 19-47-12-819.csv')
```

```python
ufo.head()
```

### Using str.contains('search_term') we can search if a particular word is present or not in a column
### The resultant is almost a boolean series since there NaN is also preserved along with True/False

```python
ufo['duration'].str.contains('hour')
```

### The resultant is almost a boolean series 

```python
ufo['city'].str.contains('Ottawa')
```

## If we try to enclose this boolean series into a dataframe then it will throw an error since there are NaN present so we need to pass na=False

```python
ufo[ufo['duration'].str.contains('hour', na=False)]
```

```python
ufo[ufo['duration'].str.contains('day|week|month', na=False)]
```
