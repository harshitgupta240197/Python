# Indexing String Series with [ ]

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
```

```python
titanic.head()
```

# We will extract the first letter from the Cabin and store in a new column called Deck

```python
titanic['cabin'].str[0]
```

```python
titanic['Deck'] = titanic['cabin'].str[0]
```

```python
titanic
```

```python
titanic.groupby('Deck').mean(numeric_only=True)
```

```python
titanic.groupby('Deck')['survived'].mean().sort_values().plot(kind='bar')
```
