# Split / Apply / Combine

### Split the data into groups based on some criteria
### Apply function(s) to each group independently
### Combine the results into a new data structure

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
titanic.head()
```

```python
titanic.info()
```

```python
titanic['age']
```

```python
titanic['age'] = titanic['age'].replace(['?'], None).astype('float')
```

```python
titanic_group = titanic.groupby('sex')['age'].mean()
titanic_group
```

```python
titanic_group['age'].mean()
```

```python
titanic_pclass_group = titanic.groupby('pclass')
```

```python
titanic_pclass_group['age'].mean()
```

```python
titanic_group_gender = titanic.groupby('sex')
titanic_group_gender
```

```python
titanic_group_gender.mean(numeric_only=True)
```
