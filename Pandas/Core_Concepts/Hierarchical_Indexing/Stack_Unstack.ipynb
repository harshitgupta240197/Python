# Stack() and Unstack()

```python
import pandas as pd
```

```python
state_pops = pd.read_csv('../../../data/state_pops.csv')
```

```python
state_pops.set_index(['state', 'year'], inplace=True)
```

```python
state_pops.head()
```

# unstack()

```python
state_pops.unstack('year').head()
```

```python
state_pops.unstack('state').head()
```

# stack()

```python
unstacked_state_pops = state_pops.unstack(level = 'state').head()
```

```python
unstacked_state_pops.stack()
```
