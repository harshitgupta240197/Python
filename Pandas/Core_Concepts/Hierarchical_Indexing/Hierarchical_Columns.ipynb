# Hierarchical Column Names

```python
import pandas as pd 
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
titanic.head()
```

```python
titanic.set_index(['pclass', 'sex'], inplace=True)
```

```python
titanic.info()
```

```python
titanic['age'] = pd.to_numeric(titanic['age'], errors='coerce').fillna(0)
```

```python
titanic['fare'] = pd.to_numeric(titanic['fare'], errors='coerce').fillna(0)
```

```python
titanic.head()
```

```python
titanic.groupby('sex').agg({
    'age' : ['min', 'max', 'mean'],
    'fare' : ['min', 'max', 'mean'],
    'survived' : ['mean']
})
```

```python
titanic_df = titanic.groupby('sex').agg({
                    'age' : ['min', 'max', 'mean'],
                    'fare' : ['min', 'max', 'mean'],
                    'survived' : ['mean']
                })
```

```python
titanic_df.index
```

#### We have got NESTED HIERARCHICAL columns

```python
titanic_df.columns
```
