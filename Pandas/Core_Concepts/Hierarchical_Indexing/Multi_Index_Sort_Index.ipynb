# Multi-Indexing - sort_index()

```python
import pandas as pd 
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
```

```python
state_pop = pd.read_csv('../../../data/state_pops.csv')
```

#### We will first set multi-index for the state population dataset using set_index()

```python
state_pop.set_index(['state', 'year'], inplace=True)
```

```python
state_pop.head()
```

## Now we will use sort_index() on this dataframe which will sort both the columns

```python
state_pop.sort_index()
```

```python
state_pop.sort_index(ascending=False)
```

# using level= we can control the column we want to sort

```python
state_pop.sort_index(ascending=False, level=0)
```

```python
state_pop.sort_index(ascending=False, level=1)
```

# We can do two differnt sorts on two different columns at once

```python
state_pop.sort_index(level = [0,1], ascending=[False,True])
```
