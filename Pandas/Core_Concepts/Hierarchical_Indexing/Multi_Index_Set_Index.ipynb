# Mutli-Indexing with set_index

```python
import pandas as pd
```

```python
state_pops = pd.read_csv('../../../data/state_pops.csv')
```

```python
state_pops.head()
```

### If we check the index of the state_pops dataframe we will see that it is a RANGE INDEX

```python
state_pops.index
```

### If we use either 'State' or 'Year' column as Index then it won't be Unique. But for Index column we want something Unique. So what we do is take combine the State and Year to make a Unique Index column

```python
state_pops.set_index(['state', 'year'])
```

```python
state_pops.set_index(['year', 'state'])
```

```python
state_pops.set_index(['year', 'state', 'population'])
```
