# Multi Index - XS Method (Cross Section)
##### It is an easier way to get/access a portion of a dataframe that is multi-indexed

```python
import pandas as pd 
```

```python
state_pops = pd.read_csv('../../../data/state_pops.csv')
state_pops.head()
```

```python
state_pops.set_index(['state', 'year'], inplace=True)
```

### We used state_pops.loc[:, 1992, :] in the previous notebook to access the contents of the second index column

## Now we will use XS method to access the content of the index column wheen there are more than one index column involved
### level= helps to define either the position of the index column or the name of the index column directly

```python
state_pops.xs(1992, level='year').head()
```

```python
state_pops.xs(2013, level=1).head()
```
