# Multi-Indexing loc[ ]

```python
import pandas as pd 
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
state_pops = pd.read_csv('../../../data/state_pops.csv')
```

```python
titanic.head()
```

##### For example I want to access the row at 7th Index position

```python
titanic.loc[7]
```

##### We can also access the entire dta frame for the 7th index position

```python
titanic.loc[[7]]
```

#### We can also search a word in a dataframe as well

```python
state_pops.head()
```

```python
state_pops.set_index(['state', 'year'], inplace=True)
```

```python
state_pops
```

```python
state_pops.loc['USA'].head()
```

#### We can also provide a list to search

```python
state_pops.loc[['AK', 'CA']]
```

#### We can also use a range to search but for that we need to sort by index first

```python
state_pops = state_pops.sort_index()
```

```python
state_pops.loc['NM' : 'TX']
```

## I want the population of Montana from 1992

```python
state_pops_MT = state_pops.loc['MT'].loc[1992]
state_pops_MT
```

##### Another way of doing this 

```python
state_pops.loc['MT', 1992]
```

##### Another safer way of doing this is by passing a tuple inside .loc[ ]

```python
state_pops.loc[('MT', 1992)]
```

## I want California's population in 2013

```python
state_pops.loc[('CA', 2013)]
```

# Using slice/range

## I want the population of Alaska from 1990 to 1995

```python
state_pops.loc[('AK', 1990) : ('AK', 1995)]
```

## I want the population of Alaska and Alabama from 2011 to 1993

```python
state_pops.loc[('AK', 2011) : ('AL', 1993)]
```

## I want to access the second index column
## df.loc[:, search_term, :]
##### : represents all of the ones before/after

```python
state_pops.loc[:, 1990, :]
```

## Using the above method we can also do a slice

```python
state_pops.loc[:, [1990, 1992], :]
```
