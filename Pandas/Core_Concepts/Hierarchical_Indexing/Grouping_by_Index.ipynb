# Grouping by Index

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
state_pops = pd.read_csv('../../../data/state_pops.csv')
```

```python
state_pops.set_index(['state', 'year'], inplace=True)
```

```python
state_pops
```

### We are finding the total population by year

```python
state_pops.groupby(level=1).sum()
```

### We will find the state with the minimum population

```python
state_pops.groupby(level=0).min().head()
```

```python
state_pops.groupby(level=[1,0]).min()
```
