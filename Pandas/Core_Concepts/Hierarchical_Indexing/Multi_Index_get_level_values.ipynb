# Multi Indexing get_level_values()
### The DataFrame.index.get_level_values() method in pandas extracts a flat vector of label values from a specific level of a multi-level index (MultiIndex). It returns a standard pandas Index object matching the exact length and order of your DataFrame rows

```python
import pandas as pd 
```

```python
state_pops = pd.read_csv('../../../data/state_pops.csv')
state_pops.head()
```

```python
state_pops.set_index(['state', 'year'], inplace=True)
```

```python
state_pops
```

```python
state_pops.index
```

```python
state_pops.index.levels
```

```python
state_pops.index.get_level_values(0)
```

```python
state_pops.index.get_level_values(1)
```

### We can also access the content inside the get_level_values() list by specifying its position

```python
state_pops.index.get_level_values(0)[20]
```

```python
state_pops.index.get_level_values(1)[900]
```

```python
state_pops.loc[:, [1990,1992,1994], :]
```

### We can also perform some maths to it and pass it as a mask into loc[ ]

```python
even_years = state_pops.index.get_level_values(1) % 2 == 0

state_pops.loc[even_years]
```

```python
state_pops[state_pops['population'] % 2 ==0]
```

### We can also access specific character of the index column

```python
state_pops.index.get_level_values(0).str[1]
```

```python
state_pops.index.get_level_values(0).str[0]
```

### We can also create a boolean condition

```python
ends_with_A = state_pops.index.get_level_values(0).str[1] == 'A'
ends_with_A
```

```python
state_pops[ends_with_A].index.values
```
