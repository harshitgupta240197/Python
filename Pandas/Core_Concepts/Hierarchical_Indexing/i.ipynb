# Multi-Indexing / Hierarchical Indexing
### Groupby with Multiple Columns

```python
import pandas as pd 
```

```python
car_stocks = pd.read_csv('../../../data/car_stocks.csv')
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
```

```python
titanic['age'] = titanic['age'].replace(['?'], None).astype('float')
titanic['flare'] = titanic['fare'].replace(['?'], None).astype('float')
```

```python
titanic.groupby('sex')['age'].mean()
```

```python
s1 = titanic.groupby('sex')['age'].mean()
s1.index
```

## Now we will attempt multi column grouping

```python
titanic.groupby(['pclass', 'sex'])['age'].mean()
```

```python
titanic.groupby(['sex', 'pclass'])['age'].mean()
```

```python
titanic_df = titanic.groupby(['pclass', 'sex']).mean(numeric_only=True)
titanic_df
```

```python
titanic_df.index
```
