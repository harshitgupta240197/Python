# Aggregations using Custom Functions

#### I want to find the difference between the highest and the lowest age by passenger class on the titanic

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
titanic
```

### Convert the age column to a float

```python
titanic['age'] = titanic['age'].replace(['?'], None).astype('float')
```

### We will first create a group with passenger class on the titanic

```python
titanic_class_group = titanic.groupby('pclass')
```

### We will create a function that gives the difference between the max and min of age

```python
def age_diff(x):
    return x.max() - x.min()
```

```python
titanic_class_group.agg({'age' : age_diff})
```

```python
salaries = pd.read_csv('../../../data/Salaries.csv', low_memory=False)
```

# Now we will load another dataet and perform custom function agg

```python
salaries.info()
```

### We will create a function to find the number of NULL values in BasePay column
#### .count() function ignores the NULL Values of a series
#### .size actually counts everything including NULL values

```python
salaries['BasePay'].count()
```

```python
salaries['BasePay'].size
```

### We will create a function that can count NULLS

```python
def null_counter(x):
    return x.size - x.count()
```

#### Now we will use our custom function 'null_counter' in a group by aggregation

```python
salaries['Year'].unique()
```

```python
salary_year_group = salaries.groupby('Year')
```

```python
salary_year_group.agg({'BasePay' : null_counter})
```
