# Aggregation in Pandas

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../../../data/titanic.csv')
titanic
```

#### Convert the age column to a float

```python
titanic['age'] = titanic['age'].replace(['?'], None).astype('float')
```

#### We will first create a group using the gender ('sex')

```python
titanic_gender_group = titanic.groupby('sex')
```

## Now we will use .agg('function')

```python
titanic_gender_group.agg('min')
```

```python
titanic_gender_group.agg('max')
```

## using .agg we can also pass a list of functions at once .agg(['function1', 'function2'])

```python
titanic_gender_group['age'].agg(['min', 'max', 'mean', 'median'])
```

## If we run a list of aggregated functions on an entire dataframe then things get chaotic

```python
titanic_gender_group.agg(['min', 'max'])
```

## So to prevent chaos we can perform column selection

```python
titanic_gender_group[['age', 'pclass']].agg(['min', 'max', 'median'])
```

## We can go one level more deep by using separate functions for seperate columns in the grouped dataframe using a dictionary

```python
titanic_gender_group.agg({'age': ['min', 'max'], 'pclass': ['median']})
```

# Now we will perform aggregation operations on the car_stocks.csv dataset

```python
car_stocks = pd.read_csv('../../../data/car_stocks.csv')
car_stocks.head()
```

```python
car_stocks.info()
```

```python
symbol_group = car_stocks.groupby('Symbol')
```

```python
symbol_group.agg({'Open' : 'mean', 'Close' : 'mean', 'Volume' : ['mean', 'sum']})
```
