# groupby()
### used to split data into groups based on specific criteria, apply a function to those groups independently, and combine the results into a structured output.

```python
import pandas as pd
```

```python
car_stocks = pd.read_csv('../../../data/car_stocks.csv')
car_stocks.head()
```

```python
LCID_df = car_stocks[car_stocks['Symbol'] == 'RIVN']

LCID_df['Close'].mean()
```

# Using groupby() we can calculate for each stock

```python
car_stocks.groupby('Symbol')['Close'].mean()
```

```python
car_stocks.groupby('Symbol')['Close'].median()
```
