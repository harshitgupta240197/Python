# Named Aggregation in Pandas
####  in pandas, named aggregation is a clean syntax used inside .groupby().agg() that allows you to calculate multiple summary statistics across different columns while explicitly naming the output columns at the same time. This prevents pandas from generating messy multi-index columns.

# ----------------------------------------------------
# df.groupby('group_column').agg(
#   new_column_name=('target_column', 'agg_func')
# )
# ----------------------------------------------------

```python
import pandas as pd 
```

```python
car_stocks = pd.read_csv('../../../data/car_stocks.csv')
```

```python
car_stocks.head()
```

#### We want to find the MIN-Open and MAX-Open and MIN-Close and MAX-Close price

```python
symbol_group = car_stocks.groupby('Symbol')
```

#### Here we will use the normal method of aggregation using a dictionary

```python
symbol_group.agg({'Open' : ['min', 'max'], 'Close' : ['min', 'max']})
```

## Now we will use NAMED AGGREGATION

```python
symbol_group.agg(
    min_open = ('Open', 'min' ),
    max_open = ('Open', 'max' ),
    min_close = ('Close', 'min'),
    max_close = ('Close', 'max')
)
```
