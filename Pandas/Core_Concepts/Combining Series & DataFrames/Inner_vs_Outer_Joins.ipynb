# Inner vs Outer Joins

```python
import pandas as pd
```

```python
fruits = pd.Series(
    data=['banana', 'orange', 'mango', 'cherry'],
    index=['b', 'o', 'm', 'c']
)

animals = pd.Series(
    data=['badger', 'owl', 'mouse', 'lamb', 'cat'],
    index=['b', 'o', 'm', 'l', 'c']
)
```

## The default JOIN is OUTER

```python
pd.concat([fruits, animals], axis = 1, keys=('fruits', 'animals'))
```

## INNER JOIN

```python
pd.concat([fruits, animals], axis=1, join='inner', keys=('fruits', 'animals'))
```
