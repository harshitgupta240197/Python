# Combining Series & Dataframes

# Concatenating Series
# pandas.concat()
#### pandas.concat() function is used to combine multiple pandas objects (DataFrames or Series) along a particular axis (either rows or columns)

```python
import pandas as pd
```

##### Let us suppose we have two series s1 and s2

```python
s1 = pd.Series(['a', 'b', 'c', 'd'])
s2 = pd.Series(['e', 'f', 'g', 'h', 'i'])
```

##### Now we will concatenate s1 and s2

```python
pd.concat([s1, s2])
```

```python
pd.concat([s2, s1])
```

### We can use ignore_index=True to generate a fresh index instead of using old index from the previous dataframe

```python
pd.concat([s2,s1], ignore_index=True)
```

### Now we will add a new pd.series that is numerical and we will try to concatenate it

```python
nums = pd.Series([1, 2, 3, 4])
```

```python
pd.concat([s1, s2, nums], ignore_index=True)
```
