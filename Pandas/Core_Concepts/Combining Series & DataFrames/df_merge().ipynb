# DataFrame merge() method
# pandas.DataFrame.merge()
#### df.merge() function in Pandas combines two DataFrames using database-style, SQL-like joins based on common columns or indexes

```python
import pandas as pd
```

```python
teams = pd.DataFrame(
    [
        ["Suns", "Phoenix", 20, 4],
        ["Mavericks", "Dallas", 11,12],
        ["Rockets", "Houston", 7, 16],
        ["Nuggets", "Denver", 11, 12]
    ],
    columns=["team", "city", "wins", "losses"]
)

cities = pd.DataFrame(
    [
        ["Houston", "Texas", 2310000],
        ["Phoenix", "Arizona", 1630000],
        ["San Diego", "California", 1410000],
        ["Dallas", "Texas", 1310000]
    ],
    columns=["city", "state", "population"]
)
```

# Inner Join (Default)

```python
teams.merge(cities)
```

# on= parameter helps us to define the field that we want to use for joining

```python
teams.merge(cities, on='city')
```

# how= parameter helps us to define the JOIN type. By default it is INNER Join

# Left Join

```python
teams.merge(cities, how='left')
```

# Right Join

```python
teams.merge(cities, how='right')
```

# Outer Join

```python
teams.merge(cities, how='outer')
```
