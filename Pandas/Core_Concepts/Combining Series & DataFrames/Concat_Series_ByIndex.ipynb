# Concatenating Series By Index
### If we do axis=1 then we can align them horizontally instead of vertically

```python
import pandas as pd
```

```python
c1 = pd.Series(['red', 'orange', 'yellow'])
c2 = pd.Series(['green', 'blue', 'purple'])
```

```python
pd.concat([c1,c2], axis=0)
```

# Using axis=1 pandas does the pairing using the INDEX as the joining part

```python
pd.concat([c1,c2], axis=1)
```

```python
fruits = pd.Series(
    data=['banana', 'orange', 'mango'],
    index=['b', 'o', 'm']
)

animals = pd.Series(
    data=['badger', 'owl', 'mouse'],
    index=['b', 'o', 'm']
)
```

```python
pd.concat([fruits, animals],axis=1)
```

## Using keys= we can rename the column

```python
pd.concat([fruits, animals],axis=1, keys=['fruit', 'animals'])
```
