```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('fivethirtyeight')
```

```python
ufos = pd.read_csv('../data/nuforc_reports.csv 19-47-12-819.csv')
ufos.head()
```

```python
ufos.info()
```

### Let us convert the date columns to date_time

```python
ufos['date_time'] = pd.to_datetime(ufos['date_time'])
ufos['posted'] = pd.to_datetime(ufos['posted'])
```

```python
ufos.info()
```

### Let us add the Year and Month columns using the date_time column

```python
ufos['year'] = ufos['date_time'].dt.year
ufos['month'] = ufos['date_time'].dt.month
```

```python
ufos.info()
```

```python
ufos['year'].value_counts().sort_index().plot(
                                            kind='line',
                                            color='red',
                                            linewidth='2',
                                            linestyle='--'
)
```

```python
ufos[ufos['state'] == 'CA'].year.value_counts().sort_index().plot(linewidth = 1.5, label = 'CA')
ufos[ufos['state'] == 'WA'].year.value_counts().sort_index().plot(linewidth = 1.5, label = 'WA')
plt.legend(loc='upper left')
```

# UFO Plotting Challenge

```python
ufos.head()
```

```python
ufos.info()
```

```python
selected_shapes = ['circle', 'light', 'triangle', 'fireball', 'formation']
selected_years = ufos['year'].between(2000,2018)
```

```python
filtered = ufos[ufos['shape'].isin(selected_shapes) & (selected_years)]
```

```python
filtered['shape']
```

```python
plt.figure(figsize=(12,5))
filtered[filtered['shape'] == 'light']['year'].value_counts().sort_index().plot(xticks=range(2000, 2019, 2), linewidth = 2, label = 'light')
filtered[filtered['shape'] == 'fireball']['year'].value_counts().sort_index().plot(xticks=range(2000, 2019, 2), linewidth = 2, label = 'fireball')
filtered[filtered['shape'] == 'circle']['year'].value_counts().sort_index().plot(xticks=range(2000, 2019, 2), linewidth = 2, label = 'circle')
filtered[filtered['shape'] == 'triangle']['year'].value_counts().sort_index().plot(xticks=range(2000, 2019, 2), linewidth = 2, label = 'triangle')
filtered[filtered['shape'] == 'formation']['year'].value_counts().sort_index().plot(xticks=range(2000, 2019, 2), linewidth = 2, label = 'formation')
plt.legend()
```
