# Box Plots
#### A box plot is a method for graphically depicting groups of numerical data through their quartiles. The box extends from the Q1 to Q3 quartile values of the data, with a line at the median (Q2). The whiskers extend from the edges of box to show the range of the data. By default, they extend no more than 1.5 * IQR (IQR = Q3 - Q1) from the edges of the box, ending at the farthest data point within that interval. Outliers are plotted as separate dots.

#### A quartile is a statistical value that splits a sorted set of data into four equal parts, each representing 25% of the total observations. The three main points are the first quartile (Q₁), the second quartile (Q₂), and the third quartile (Q₃)

```python
import pandas as pd 
import matplotlib.pyplot as plt

plt.style.use('fivethirtyeight')
```

```python
salaries = pd.read_csv('../data/Salaries.csv', low_memory=False)
```

```python
salaries.head()
```

```python
salaries.info()
```

#### We will convert a few columns to numeric

```python
salaries['BasePay'] = pd.to_numeric(salaries['BasePay'], errors='coerce').fillna(0)
salaries['OvertimePay'] = pd.to_numeric(salaries['OvertimePay'], errors='coerce').fillna(0)
salaries['OtherPay'] = pd.to_numeric(salaries['OtherPay'], errors='coerce').fillna(0)
```

```python
salaries_df = salaries[['BasePay', 'OvertimePay', 'OtherPay']]
salaries_df.head()
```

```python
salaries_df.plot(kind='box')
```

#### Another method of generating the same box plt

```python
salaries_df.boxplot()
```

```python
houses = pd.read_csv('../data/kc_house_data.csv')
houses.head()
```

# To control show/hide of OUTLIER Values we can use 
# .plot(showfliers=False)
