```python
import pandas as pd 
import matplotlib.pyplot as plt

plt.style.use('fivethirtyeight')
```

```python
ufo = pd.read_csv('../data/nuforc_reports.csv 19-47-12-819.csv')
```

```python
ufo.info()
```

```python
ufo['date_time'] = pd.to_datetime(ufo['date_time'])
```

```python
ufo['month'] = ufo['date_time'].dt.month
```

```python
ufo.head()
```

```python
sightings_by_month = ufo['month'].value_counts().sort_index()
sightings_by_month
```

```python
sightings_by_month.rename({
    1: 'Jan',
    2: 'Feb',
    3: 'Mar',
    4: 'Apr',
    5: 'May',
    6: 'Jun',
    7: 'Jul',
    8: 'Aug',
    9: 'Sep',
    10: 'Oct',
    11: 'Nov',
    12: 'Dec',
}, inplace=True)
```

```python
sightings_by_month.plot(kind='bar', title='UFO Sightings by month')

plt.xlabel('Month', fontsize = 14)
plt.ylabel('# of sightings', fontsize=14)
```
