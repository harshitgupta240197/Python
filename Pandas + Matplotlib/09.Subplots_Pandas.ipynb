# Subplots

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
salary = pd.read_csv('../data/Salaries.csv')
salary.head()
```

```python
salary.info()
```

```python
salary['BasePay'] = pd.to_numeric(salary['BasePay'], errors='coerce').fillna(0)
salary['OvertimePay'] = pd.to_numeric(salary['OvertimePay'], errors='coerce').fillna(0)
salary['OtherPay'] = pd.to_numeric(salary['OtherPay'], errors='coerce').fillna(0)
```

```python
salary.info()
```

```python
salary_df = salary[['BasePay', 'OvertimePay', 'OtherPay']]
salary_df.head()
```

```python
salary_df.plot(
        kind = 'hist',
        subplots = True, 
        figsize = (5,5),
        sharex = False,
        bins = 40)
plt.tight_layout()
```

### To control the horizontal and vertical display we can use layout=(rows,columns)

```python
salary_df.plot(
        kind = 'hist',
        subplots = True, 
        figsize = (10,3),
        sharex = False,
        layout = (1,3), 
        bins = 40)
plt.tight_layout()
```

# Accessing INDIVIDUAL PLOTS

#### First we will store the entire plot inside a variable and call it axs

#### Now we can access this variable components using [ ]

```python
axs = salary_df.plot(
        kind = 'hist',
        subplots = True, 
        figsize = (10,3),
        sharex = False,
        layout = (1,3), 
        bins = 40)

axs[0][0].set_title('Base Pay')
axs[0][1].set_title('Overtime Pay')
axs[0][2].set_title('Other Pay')

axs[0][0].set_xlim(0,250000)
axs[0][1].set_xlim(0,90000)
axs[0][2].set_xlim(0,50000)

plt.tight_layout()
```
