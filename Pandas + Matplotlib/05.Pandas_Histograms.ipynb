# Pandas HISTOGRAMS - 
# .plot(kind = 'hist')

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
salaries = pd.read_csv('../data/Salaries.csv', low_memory=False)
salaries.head()
```

```python
salaries.info()
```

```python
salaries['BasePay'] = pd.to_numeric(salaries['BasePay'], errors='coerce').fillna(0)
salaries['OvertimePay'] = pd.to_numeric(salaries['OvertimePay'], errors='coerce').fillna(0)
salaries['OtherPay'] = pd.to_numeric(salaries['OtherPay'], errors='coerce').fillna(0)
```

```python
salaries.info()
```

```python
salaries['BasePay'].plot(
                        kind='hist', 
                        bins=200,
                        color=)
```

## We can plot the histogram for the entire dataframe as well

```python
salaries.plot(kind='hist', alpha=0.5, bins = 20)
```

# df.hist()
### Using this it will make a histogram for every piece in that dataframe

```python
salaries_df = salaries[['BasePay', 'OvertimePay', 'OtherPay']]
```

```python
salaries_df.hist()
```
