```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
ufo = pd.read_csv('../data/nuforc_reports.csv 19-47-12-819.csv')
ufo.head()
```

```python
ufo.info()
```

#### Converting the date_time column to date_time format

```python
ufo['date_time'] = pd.to_datetime(ufo['date_time'])
```

#### Creating a new month column using the date_time column

```python
ufo['month'] = ufo['date_time'].dt.month
```

```python
ufo.head()
```

## We are using Pandas only here

```python
ufo['month'].value_counts().sort_index().plot(
                                            kind='bar', 
                                            title='UFO Sightings by month',
                                            xlabel='Month',
                                            ylabel='# of sightings')
```

## Now we will use Pandas + Matplotlib
## Matplotlib gives more control over the plot as compared to pandas

```python
ufo['month'].value_counts().sort_index().plot(
                                            kind='bar', 
                                            title='UFO Sightings by month',
                                            )
plt.xlabel('Month', fontsize = 14)
plt.ylabel('# of sightings', fontsize=14)
```
