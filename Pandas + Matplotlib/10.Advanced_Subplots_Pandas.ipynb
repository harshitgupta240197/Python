# Advanced SUBPLOTS in pandas

```python
import pandas as pd 
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
ufo  = pd.read_csv('../data/nuforc_reports.csv 19-47-12-819.csv')
```

```python
ufo.head()
```

```python
ufo.info()
```

```python
ufo['date_time'] = pd.to_datetime(ufo['date_time'])
```

```python
ufo['month'] = ufo['date_time'].dt.month
ufo['year'] = ufo['date_time'].dt.year
```

```python
ufo_2014 = ufo[ufo['year'] == 2014]
ufo_2015 = ufo[ufo['year'] == 2015]
ufo_2016 = ufo[ufo['year'] == 2016]
ufo_2017 = ufo[ufo['year'] == 2017]
ufo_2018 = ufo[ufo['year'] == 2018]
ufo_2019 = ufo[ufo['year'] == 2019]
```

```python
months = {1:'Jan', 2:'Feb', 3:'Mar', 4:'Apr', 5:'May', 6:'Jun', 7:'Jul', 8:'Aug',9:'Sep',10:'Oct',11:'Nov',12:'Dec'}
fig, axes = plt.subplots(2,3, figsize = (15,8))
plt.suptitle('UFO Sightings by month')

ufo_2014['month'].value_counts().sort_index().rename(months).plot(kind='bar', ax=axes[0,0], color = '#7393B3', title='2014')
ufo_2015['month'].value_counts().sort_index().rename(months).plot(kind='bar', ax=axes[0,1], color = '#7393B3', title='2015')
ufo_2016['month'].value_counts().sort_index().rename(months).plot(kind='bar', ax=axes[0,2], color = '#7393B3', title='2016')
ufo_2017['month'].value_counts().sort_index().rename(months).plot(kind='bar', ax=axes[1,0], color = '#7393B3', title='2017')
ufo_2018['month'].value_counts().sort_index().rename(months).plot(kind='bar', ax=axes[1,1], color = '#7393B3', title='2018')
ufo_2019['month'].value_counts().sort_index().rename(months).plot(kind='bar', ax=axes[1,2], color = '#7393B3', title='2019')

plt.tight_layout()
```
