```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
titanic = pd.read_csv('../../data/titanic.csv')
titanic.head()
```

```python
titanic.info()
```

#### First we will convert the 'age' column to numeric

```python
titanic['age'] = pd.to_numeric(titanic['age'], errors='coerce')
```

```python
titanic['pclass'].unique()
```

```python
first_class = titanic[titanic['pclass'] == 1]
second_class = titanic[titanic['pclass'] == 2]
third_class = titanic[titanic['pclass'] == 3]
```

```python
plt.figure(figsize=(12,4))
plt.suptitle('Titanic Age Histogram')

plt.subplot(1,3,1)
plt.hist(
    first_class['age'], 
    label= '1st class', 
    color='blue',
    bins=20)
plt.title('1st class')

plt.subplot(1,3,2)
plt.hist(
    second_class['age'], 
    label='2nd class', 
    color= 'orange',
    bins=20)
plt.title('2nd class')

plt.subplot(1,3,3)
plt.hist(
    third_class['age'], 
    label= '3rd class', 
    color= 'red',
    bins=20)
plt.title('3rd class')

plt.tight_layout()
```
