# Subplots in Matplotlib - matplotlib.plt.subplot()
#### Subplot expects us to input # of rows, # of columns, index

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.style.use('fivethirtyeight')
```

```python
nums = np.arange(5)
```

#### .suptitle() - supertitle
#### .tight_layout() - automatically adjusts subplot padding and margins

```python
plt.figure(figsize=(10,4))
plt.suptitle('3 Charts Subplot')   

plt.subplot(1,3,1)
plt.plot(nums, nums)
plt.title('1st Chart')

plt.subplot(1,3,2)
plt.plot(nums, nums ** 3, color = 'orange')
plt.title('2nd Chart')

plt.subplot(1,3,3)
plt.plot(nums, nums ** 2, color = 'red')
plt.title('3rd Chart')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(4,6))
plt.suptitle('3 Charts Subplot')   

plt.subplot(3,1,1)
plt.plot(nums, nums)
plt.title('1st Chart')

plt.subplot(3,1,2)
plt.plot(nums, nums ** 3, color = 'orange')
plt.title('2nd Chart')

plt.subplot(3,1,3)
plt.plot(nums, nums ** 2, color = 'red')
plt.title('3rd Chart')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.suptitle('3 Charts Subplot')   

plt.subplot(2,2,1)
plt.plot(nums, nums)
plt.title('1st Chart')

plt.subplot(2,2,2)
plt.plot(nums, nums ** 3, color = 'orange')
plt.title('2nd Chart')

plt.subplot(2,2,3)
plt.plot(nums, nums ** 2, color = 'red')
plt.title('3rd Chart')

plt.subplot(2,2,4)
plt.plot(nums, nums ** 4, color = 'brown')
plt.title('4th Chart')

plt.tight_layout()
plt.show()
```
