# Matplotlib Legends

```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
nums = np.arange(5)
```

```python
plt.style.use('ggplot')
```

### Now we will use plt.legend() to show the legend. We have to make sure that labels are active before activating legend

```python
plt.figure()
plt.plot(nums, nums, label = 'x')
plt.plot(nums, nums * nums, label = 'x-squared')
plt.plot(nums, nums ** 3, label = 'x-cubed')
plt.legend()
```

### We can control the position of the legend using the loc= parameter

```python
plt.figure()
plt.plot(nums, nums, label = 'x')
plt.plot(nums, nums * nums, label = 'x-squared')
plt.plot(nums, nums ** 3, label = 'x-cubed')
plt.legend(loc= 'upper left')
```
