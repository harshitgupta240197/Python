# Matplotlib - Line Styles (Color, Width...)

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
nums = np.arange(5)
```

# Line color

### We can use the color parameter to change the color of the lines on line chart and we can also use hex code

```python
plt.figure()
plt.plot(nums, nums, color = 'purple')
plt.plot(nums, nums * nums, color = 'blue')
plt.plot(nums, nums ** 3)
```

# Line Width

### We can use the Line Width property to change the line width

```python
plt.figure()
plt.plot(nums, nums, linewidth = 3)
plt.plot(nums, nums * nums, linewidth = 3)
plt.plot(nums, nums ** 3, linewidth = 3)
```

# Line Style

```python
plt.figure()
plt.plot(nums, nums, linestyle = 'dashed')
plt.plot(nums, nums * nums, linestyle = 'dotted')
plt.plot(nums, nums ** 3, linestyle = '--')
plt.plot(nums, nums ** 4, linestyle = '-.')
```

# Line Marker
#### Using line markers and controlling its size and color

```python
plt.figure()
plt.plot(nums, nums, marker = '*', markersize = 8, markerfacecolor = 'pink')
plt.plot(nums, nums * nums, marker = '>', markersize = 12, markerfacecolor = 'orange')
plt.plot(nums, nums ** 3, marker = '.', markersize = 20, markerfacecolor = 'yellow')
```
