# matplotlib.pyplot.scatter() - Scatter Plot

```python
import matplotlib.pyplot as plt
```

```python
plt.style.use('fivethirtyeight')
```

```python
heights = [137,140,142,145,147,150,152,155,157,160]
f_weights = [28.5,30.8,32.6,34.9,36.4,39,40.8,43.1,44.9,47.2]
m_weights = [34.9,38.1,33.5,35.8,46.7, 42.8,43.1,45.8,50.8,58.9]
```

```python
plt.scatter(heights, f_weights, marker= '*', color = 'purple', label = 'female weights')
plt.scatter(heights, m_weights, marker= '*', color = 'blue', label = 'male weights')
plt.legend()
plt.xlabel('Height')
plt.ylabel('Weight')
```
