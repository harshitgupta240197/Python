# X and Y Ticks

```python
import matplotlib.pyplot as plt
```

```python
plt.style.use('fivethirtyeight')
```

### We will use both xticks and yticks in the below example

```python
salaries = [23000,25000,21000,20000,30000]
age = [20,24,40,60,65]

plt.plot(age, salaries)
plt.title('Employee Salaries')
plt.xlabel('age')
plt.ylabel('salaries')
plt.xticks([20,30,40,50,60,70])
plt.yticks([20000,25000,30000,35000])
```

### We can also configure the TICK LABELS

```python
salaries = [23000,25000,21000,20000,30000]
age = [20,24,40,60,65]

plt.plot(age, salaries)
plt.title('Employee Salaries')
plt.xlabel('age')
plt.ylabel('salaries')
plt.xticks([20,30,40,50,60,70])
plt.yticks([20000,25000,30000,35000], labels=['20K', '25K', '30K', '35K'])
```
