## Now we will see the age distribution for the passengers on the titanic

```python
titanic = pd.read_csv('../../data/titanic.csv')
titanic.head()
```

```python
titanic.info()
```

```python
titanic['age'] = pd.to_numeric(titanic['age'], errors='coerce')
```

```python
plt.hist(titanic['age'], bins = 50)
```

### Let us see the age distribution for the First Class passengers on titanic

```python
first_class = titanic[titanic['pclass'] == 1]
first_class.head()
```

```python
plt.hist(first_class['age'])
```

### Let us plot the age distribution for the first and second class passengers on the titanic

# To control Transparency we use alpha=

```python
first_class = titanic[titanic['pclass'] == 1]
second_class = titanic[titanic['pclass'] == 2]

plt.hist(first_class['age'], label = '1st class')
plt.hist(second_class['age'], label = '2nd class', alpha = 0.6)

plt.legend()
plt.title('Titanic Age Distribution')
plt.xlabel('Age')
```
