# Plot Labels and Titles in Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
plt.style.use('seaborn-v0_8')
```

# Chart Title
#### plt.title() is order sensitive

```python
salaries = [23000,25000,21000,20000,30000]
age = [20,24,40,60,65]

plt.plot(age, salaries)
plt.title('Employee Salaries')
```

### Chart Title Alignment using loc = ('left/right')

```python
salaries = [23000,25000,21000,20000,30000]
age = [20,24,40,60,65]

plt.plot(age, salaries)
plt.title('Employee Salaries', loc='left')
```

# Chart Labels
## xlabel() and ylabel() and their related properties

```python
salaries = [23000,25000,21000,20000,30000]
age = [20,24,40,60,65]

plt.plot(age, salaries)
plt.title('Employee Salaries', loc='left')
plt.xlabel('age')
plt.ylabel('salaries')
```

### For label based padding/space we use labelpad=

```python
salaries = [23000,25000,21000,20000,30000]
age = [20,24,40,60,65]

plt.plot(age, salaries)
plt.title('Employee Salaries', loc='left')
plt.xlabel('age', labelpad=20)
plt.ylabel('salaries')
```

# Label and Ttile Font Size

```python
salaries = [23000,25000,21000,20000,30000]
age = [20,24,40,60,65]

plt.plot(age, salaries)
plt.title('Employee Salaries', loc='left', fontsize = 24)
plt.xlabel('age', labelpad=20, fontsize = 24)
plt.ylabel('salaries', fontsize = 24)
```

# Label Color

```python
salaries = [23000,25000,21000,20000,30000]
age = [20,24,40,60,65]

plt.plot(age, salaries)
plt.title('Employee Salaries', loc='left', fontsize = 24, color = 'blue')
plt.xlabel('age', labelpad=20, fontsize = 24, color = 'blue')
plt.ylabel('salaries', fontsize = 24, color = 'blue')
```
