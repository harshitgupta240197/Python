# Matplotlib BAR Plots - matplotlib.pyplot.bar()

```python
import matplotlib.pyplot as plt
```

```python
plt.style.use('ggplot')
```

```python
plants = ['kale', 'spinach', 'broccoli', 'Turnip']
died = [10,25,5,30]
germinated = [78,56,40,59]
```

```python
plt.figure(figsize=(6,6))
plt.bar(plants, died, bottom=germinated, label = 'Died')
plt.bar(plants, germinated, label = 'Germinated')
plt.title('Plant Germination Rate')
```
