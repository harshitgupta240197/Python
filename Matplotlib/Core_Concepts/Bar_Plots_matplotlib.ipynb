# Matplotlib BAR Plots - matplotlib.pyplot.bar()

```python
import matplotlib.pyplot as plt
```

```python
plt.style.use('ggplot')
```

```python
plants = ['kale', 'spinach', 'broccoli', 'Turnip']
died = [10,25,5,30]
germinated = [78,56,40,59]
```

#### Creating a bar plot

##### Stacked Bar Chart

```python
plt.figure(figsize=(6,6))
plt.bar(plants, died, bottom=germinated, label = 'Died', color = 'blue')
plt.bar(plants, germinated, label = 'Germinated', color = 'green')
plt.title('Plant Germination Rate')
plt.ylabel('Number of seedlings')
```

```python
plt.figure(figsize=(6,6))
plt.bar(plants, died, label = 'Died', color = 'blue')
plt.bar(plants, germinated, label = 'Germinated', color = 'green', width = 0.4)
plt.title('Plant Germination Rate')
plt.ylabel('Number of seedlings')
plt.legend(loc = 'upper left')
```

### Horizontal Bar Chart

```python
plt.figure(figsize=(6,6))
plt.barh(plants, died, label = 'Died', color = 'blue')
plt.barh(plants, germinated, label = 'Germinated', color = 'green', height = 0.4)
plt.title('Plant Germination Rate')
plt.ylabel('Number of seedlings')
plt.legend(loc = 'upper left')
```
