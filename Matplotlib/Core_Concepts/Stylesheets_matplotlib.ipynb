# Matplotlib Stylesheets

```python
import matplotlib.pyplot as plt
import numpy as np
```

### We can check which all styles are available for us

```python
plt.style.available
```

### Out of the list we will use ''fivethirtyeight''

```python
plt.style.use('fivethirtyeight')
```

```python
nums = np.arange(5)
```

```python
plt.figure()
plt.plot(nums, nums)
plt.plot(nums ** 3)
plt.plot(nums * nums)
```

### This time we will use 'seaborn-v0_8' style sheet

```python
plt.style.use('seaborn-v0_8')
```

```python
plt.figure()
plt.plot(nums, nums)
plt.plot(nums ** 3)
plt.plot(nums * nums)
```
