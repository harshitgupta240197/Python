# Pie Charts - matplotlib.pyplot.pie()

```python
import matplotlib.pyplot as plt
```

```python
plt.style.use('fivethirtyeight')
```

```python
labels = ['Tomato', 'Onions', 'Potatoes', 'Chillies']
prices = [25.99, 3.24, 9.50, 6.99]
```

## plt.pie(size, labels)

```python
plt.pie(prices, labels=labels)
plt.legend(loc = 'upper left')
plt.show()
```

## autopct = '%1.1f%%' - Percentage

```python
plt.pie(prices, labels=labels, autopct = '%1.1f%%')
plt.show()
```

## Using pctdistance= we can adjust the distance of the percentage labels

```python
plt.pie(prices, labels=labels, autopct = '%1.1f%%', pctdistance=1.3)
plt.show()
```

## We can put a SHADOW= as well to the pie chart

```python
plt.pie(prices, labels=labels, autopct = '%1.1f%%', pctdistance=1.3, shadow=True)
plt.show()
```

## We can also specify the ANGLE at which the pie chart has to be rotated using startangle=

```python
plt.pie(prices, labels=labels, autopct = '%1.1f%%', shadow=True, startangle=90)
plt.show()
```

## EXPLODE - offsetting a slice using explode

```python
explode = (0,0,0.1,0)

plt.pie(prices, labels=labels, autopct = '%1.1f%%', shadow=True, explode=explode)
plt.show()
```
