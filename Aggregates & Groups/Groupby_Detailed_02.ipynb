# groupby() Detailed and its methods

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../data/titanic.csv')
titanic.head()
```

```python
titanic.info()
```

```python
titanic['age']
```

### We will convert the 'age' column to float since it is an object now

```python
titanic['age'] = titanic['age'].replace(['?'], None).astype('float')
```

```python
titanic.info()
```

```python
titanic_df = titanic[['pclass', 'survived', 'sex', 'age']]
titanic_df.head()
```

## We are doing a groupby() by the 'sex' column

```python
new_groupby = titanic_df.groupby('sex')
new_groupby
```

# .ngroups gives us the # of groups in this groupby object

```python
new_groupby.ngroups
```

# using .groups we can get the mapping/dictionary of the groupby object

```python
new_groupby.groups
```

# To see the first member of a group we use .first()

```python
new_groupby.first()
```

# We can use .get_group() method to access an individual group

```python
new_groupby.get_group('female')
```

```python
new_groupby.get_group('male')
```

# We can iterate over a group using FOR Loop

```python
for name, group in new_groupby:
    print(name)
    print('--------------------------------------')
    print(group)
```
